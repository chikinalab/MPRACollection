# Download MPRA datasets from the command line

Installing MPRACollection also installs the `mdc` command, which downloads datasets from the
[MPRA Dataset Collection repo on HuggingFace](https://huggingface.co/datasets/jnhu323/MPRA_Dataset_Collection).

| Command | What it does |
|---|---|
| `mdc list_dataset` | List all papers and their datasets |
| `mdc download_dataset [DATA_DIR] --paper P --dataset D` | Download one dataset |
| `mdc download_dataset [DATA_DIR] --paper P` | Download all datasets of one paper |
| `mdc download_dataset [DATA_DIR] --all` | Download all datasets of all papers |
| `mdc load_dataset [DATA_DIR] --paper P --dataset D` | Load one dataset (downloading it first if needed) and print a summary |

- `--paper` / `-p` and `--dataset` / `-d` take the names shown by `mdc list_dataset`. `--dataset` always needs `--paper`.
- **DATA_DIR** is optional. If you leave it out, the files are saved in the HuggingFace cache (`~/.cache/huggingface/hub` by default).
- Add `--overwrite` to `download_dataset` to download files again even if they already exist.

> In this notebook, the `!` at the start of a cell runs it as a shell command. In a terminal, type the commands without the `!`.

```python
!mdc --help
```

## 1. See what is available

`mdc list_dataset` prints every paper in the repo with its datasets. Use these names for `--paper` and `--dataset`.

```python
!mdc list_dataset
```

## 2. Download datasets

Each dataset comes as two files: `<dataset>.csv.gz` (sequences and readouts, gzip-compressed) and `<dataset>.yaml` (metadata of this table).
Each paper also has a `study.yaml` (metadata shared by all datasets of the paper), which is downloaded together with any of its datasets.
Files that are already downloaded and up to date are not downloaded again.

### One dataset

Give both `--paper` and `--dataset`. Without DATA_DIR, the files go into the HuggingFace cache.

```python
!mdc download_dataset --paper eLife_2020_King --dataset MiniSpacer
```

Give a folder as DATA_DIR to save the files there instead, as `DATA_DIR/<paper>/<dataset>.csv.gz`, `DATA_DIR/<paper>/<dataset>.yaml` and `DATA_DIR/<paper>/study.yaml`. The folder is created if it does not exist. `-p` and `-d` are short for `--paper` and `--dataset`.

```python
!mdc download_dataset data -p eLife_2020_King -d MiniSpacer
!ls data/eLife_2020_King
```

### All datasets of one paper

Leave out `--dataset`.

```python
!mdc download_dataset data --paper eLife_2020_King
```

### All datasets of all papers

Use `--all` (without `--paper`). This downloads the whole repo, **about 7.7 GB**

### Download again

`--overwrite` downloads the files even if they already exist, e.g. to replace files you have changed locally. It works with all of the above.

```python
!mdc download_dataset data -p eLife_2020_King -d MiniSpacer --overwrite
```

## 3. Load a dataset

`mdc load_dataset` loads the dataset (downloads it first if needed), and prints a summary. The summary includes the major meta information. Full information can be found at dataset-specific yaml file and `study.yaml`.

```python
!mdc load_dataset data -p eLife_2020_King -d MiniSpacer
```

## 4. Use the dataset in Python

The command line only prints a summary. To work with the data, load it in Python:

- `resources.load_dataset(paper, dataset, local_dir)` does the same as `mdc load_dataset` and returns an `MPRA_Dataset`. The full metadata is in `mpra_dataset.info_study` and `mpra_dataset.info`.
- `MPRA_Dataset.load(paper, dataset, folder)` loads files that are already in `folder`, without going online.

```python
from MPRACollection import resources, MPRA_Dataset

mpra_dataset = resources.load_dataset("eLife_2020_King", "MiniSpacer", local_dir="data")
mpra_dataset.Y.head()
```
