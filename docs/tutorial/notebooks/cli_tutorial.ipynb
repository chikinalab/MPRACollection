# Download MPRA datasets from the command line

Installing MPRACollection also installs the `mdc` command, which downloads datasets from the
[MPRACollection repo on HuggingFace](https://huggingface.co/datasets/simon1727/MPRACollection).

| Command | What it does |
|---|---|
| `mdc list_dataset` | List all papers and their datasets |
| `mdc download_dataset [DATA_DIR] --paper P --dataset D` | Download one dataset |
| `mdc download_dataset [DATA_DIR] --paper P` | Download all datasets of one paper |
| `mdc download_dataset [DATA_DIR] --all` | Download all datasets of all papers |
| `mdc load_dataset [DATA_DIR] --paper P --dataset D` | Load one dataset (downloading it first if needed) and print a summary |

- `--paper` / `-p` and `--dataset` / `-d` take the names shown by `mdc list_dataset`. `--dataset` always needs `--paper`.
- **DATA_DIR** is optional. If you leave it out, the files are saved in the HuggingFace cache (`~/.cache/huggingface/hub` by default).
- Add `--overwrite` to `download_dataset` to download files again even if they already exist.

> In this notebook, the `!` at the start of a cell runs it as a shell command. In a terminal, type the commands without the `!`.

```python
!mdc --help
```

## 1. See what is available

`mdc list_dataset` prints every paper in the repo with its datasets. Use these names for `--paper` and `--dataset`.

```python
!mdc list_dataset
```

## 2. Download datasets

Each dataset comes as two files: `<dataset>.csv` (sequences and readouts) and `<dataset>.yaml` (metadata).
Files that are already downloaded and up to date are not downloaded again.

### One dataset

Give both `--paper` and `--dataset`. Without DATA_DIR, the files go into the HuggingFace cache.

```python
!mdc download_dataset --paper NatureGenetics_2023_Cohen --dataset state
```

Give a folder as DATA_DIR to save the files there instead, as `DATA_DIR/<paper>/<dataset>.csv` and `.yaml`. The folder is created if it does not exist. `-p` and `-d` are short for `--paper` and `--dataset`.

```python
!mdc download_dataset data -p NatureGenetics_2023_Cohen -d state
!ls data/NatureGenetics_2023_Cohen
```

### All datasets of one paper

Leave out `--dataset`.

```python
!mdc download_dataset data --paper NatureGenetics_2023_Cohen
```

### All datasets of all papers

Use `--all` (without `--paper`). This downloads the whole repo, **about 27 GB**

### Download again

`--overwrite` downloads the files even if they already exist, e.g. to replace files you have changed locally. It works with all of the above.

```python
!mdc download_dataset data -p NatureGenetics_2023_Cohen -d state --overwrite
```

## 3. Load a dataset

`mdc load_dataset` downloads the dataset if needed, loads it, and prints a summary: the number of sequences and readouts, the column names and the metadata.
It is a quick way to check a dataset before using it in Python. It needs both `--paper` and `--dataset`.

```python
!mdc load_dataset data -p NatureGenetics_2023_Cohen -d state
```

As with `download_dataset`, leaving out DATA_DIR uses the HuggingFace cache.

```python
!mdc load_dataset -p NatureGenetics_2023_Cohen -d state
```

## 4. Use the dataset in Python

The command line only prints a summary. To work with the data, load it in Python:

- `datasets.load_dataset(paper, dataset, local_dir)` does the same as `mdc load_dataset` and returns an `MPRA_Dataset`.
- `MPRA_Dataset.load(paper, dataset, folder)` loads files that are already in `folder`, without going online.

See [test_load_data.ipynb](test_load_data.ipynb) for what you can do with an `MPRA_Dataset`.

```python
from MPRACollection import datasets, MPRA_Dataset

mpra_dataset = datasets.load_dataset("NatureGenetics_2023_Cohen", "state", local_dir="data")
# same as: MPRA_Dataset.load("NatureGenetics_2023_Cohen", "state", "data")
mpra_dataset.Y.head()
```

## Common mistakes

- **`--dataset` without `--paper`**: a dataset name is only meaningful within its paper, so always give `--paper` too.
- **Paper or dataset not found**: names are case-sensitive, check them with `mdc list_dataset`.
